### **Exploración Scrap Perfil de Docentes**

#### ***Load and transform data***

```python
import pandas as pd 
from pathlib import Path

### Load Data
DATA_PATH = Path("../scripts/docentes_industrial_sistemas.csv")
perfiles = pd.read_csv(DATA_PATH)

### Transform Data
perfiles.columns = perfiles.columns.str.lower()
perfiles['perfil'] = perfiles['perfil'].str.lower() 
```

#### ***Utils***

```python
def filter_profile_by_word(df:pd.DataFrame,word:str):
    """
    Función para filtrar registros por coincidencia de termino en el perfil del docente. 
    """
    NEGRITA = "\033[1m"
    RESET = "\033[0m"
    output_df = df[df['perfil'].str.contains(word)].reset_index(drop=True)
    print('Información de los docentes:\n')
    if output_df.empty == True:
        print('No hay conincidencias. :(')
    else:

        for row in range(len(output_df)):
            print(f'\n{NEGRITA}{output_df.loc[row,'nombre']} - {output_df.loc[row,'departamento']}{RESET}')
            print(f'{NEGRITA}Cargo:{RESET} {output_df.loc[row,'cargo']}')
            print(f'{NEGRITA}Formación academica:{RESET} {output_df.loc[row,'formacion_academica']}')
        
            print(f'{NEGRITA}Perfil:{RESET} {output_df.loc[row,'perfil']}')
            print(f'{NEGRITA}URL:{RESET} {output_df.loc[row,'url_perfil']}')

            print('----------'*10)
    return output_df
```
